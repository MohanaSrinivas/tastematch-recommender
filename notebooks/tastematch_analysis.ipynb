# TasteMatch exploratory analysis\nThis notebook loads MovieLens latest-small, inspects the data, creates a per-user temporal split, fits two recommenders and reports ranking metrics. Results are computed when executed; none are pre-filled.

```python
import sys\nsys.path.insert(0, '../src')\nfrom tastematch.data import load_movielens\nfrom tastematch.split import temporal_user_split\nfrom tastematch.content import ContentRecommender\nfrom tastematch.collaborative import ItemKNNRecommender\nfrom tastematch.evaluation import evaluate_users
```

```python
ratings, movies = load_movielens()\nratings.head(), movies.head(), ratings.shape, movies.shape
```

```python
train, test = temporal_user_split(ratings)\ntrain.shape, test.shape, test.groupby('userId').size().describe()
```

```python
content = ContentRecommender().fit(movies)\ncollaborative = ItemKNNRecommender().fit(train)\ncontent_metrics = evaluate_users(test, lambda u,k: content.recommend(int(test[test.userId == u].iloc[0].movieId), k))\ncollaborative_metrics = evaluate_users(test, collaborative.recommend)\ncontent_metrics, collaborative_metrics
```
